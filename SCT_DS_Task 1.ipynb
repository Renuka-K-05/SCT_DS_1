<h4>To Find the which numeric column is completely empty</h4>

```python
lst=[]
for col in data.select_dtypes(include="float64").columns:
    if data[col].isnull().sum()== len(data):
        print(f"\n{col}:")
        print(data[col].value_counts())
        lst.append(col)
print("columns with overall empty values:", lst)

    
```

```python
for col in lst:
    data.drop(col,axis=1,inplace=True)
```

<h2>Visualizing the plots</h2>

<h3>Bar Chart Observations</h3>

-population distribution is highly uneven across countries.<br>
-A small number if countries account for a large share of the global population.<br>
-Most countries have comparitively smaller populations.<br>
-The bar chart highlights the countries with the highest populations.<br>

```python
countries=data[data['Country Code'].str.len()==3]
top10= countries.sort_values(by="2024" , ascending=False).head(10)

colors=plt.cm.viridis(np.linspace(0,1,len(top10)))
plt.figure(figsize=(10,6))
plt.bar(top10['Country Name'],top10['2024'],color=colors)
plt.xticks(rotation=80)
plt.xlabel("Country")
plt.ylabel("Population")
plt.title("Top 10 Most populated Countries in 2024")
plt.tight_layout()
plt.show()
```
